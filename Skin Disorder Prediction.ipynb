```python
# --- Machine Learning Models Initialization ---

models = {

    "Logistic Regression":
        LogisticRegression(max_iter=1000, random_state=42),

    "Decision Tree":
        DecisionTreeClassifier(random_state=42),

    "Random Forest":
        RandomForestClassifier(random_state=42),

    "Gradient Boosting":
        GradientBoostingClassifier(),

    "Extra Trees":
        ExtraTreesClassifier(random_state=42),

    "Support Vector Machine":
        SVC(probability=True),

    "K-Nearest Neighbors":
        KNeighborsClassifier(),

    "Naive Bayes":
        GaussianNB(),

    "XG Boost":
        XGBClassifier()
}

print("\n✅ All Models Initialized Successfully!")
```

#### Insight:
Multiple machine learning models were initialized to compare classification performance for skin disorder prediction.

Different algorithms learn data patterns using different approaches, allowing performance comparison across 
linear,

```python
# --- Model Training and Evaluation ---

results = {}

for model_name, model in models.items():

    # --- Train the Model ---
    
    model.fit(X_train, y_train)

    # --- Predictions ---
    
    pred = model.predict(X_test)

    # --- Accuracy ---
    
    acc = accuracy_score(y_test, pred)

    # --- Precision ---
    
    prec = precision_score(y_test,pred,average='weighted',zero_division=0)

    # --- Recall ---
    
    rec = recall_score( y_test, pred, average='weighted', zero_division=0)

    # --- F1 Score ---
    
    f1 = f1_score(y_test, pred, average='weighted', zero_division=0)

    # --- Cross Validation Score ---
    
    cv_score = cross_val_score(model, X_scaled, y, cv=5,scoring='accuracy').mean()

    # --- Training Accuracy ---
    
    train_acc = accuracy_score( y_train, model.predict(X_train))

    # --- Testing Accuracy ---
    
    test_acc = model.score(X_test, y_test)

    # --- Store Results ---
    
    results[model_name] = {

        "Accuracy": acc,

        "Precision": prec,

        "Recall": rec,

        "F1 Score": f1,

        "CV Score": cv_score,

        "Train Accuracy": train_acc,

        "Test Accuracy": test_acc
    }

print("\n✅ Model Training and Evaluation Completed Successfully")
```

#### Insight:
Multiple machine learning models were trained and evaluated to compare classification performance for skin disorder prediction.

Evaluation metrics such as Accuracy, Precision, Recall, F1 Score, ROC-AUC, and Cross Validation Score help measure model reliability, classification capability, and generalization performance.

Comparing multiple models helps identify the most accurate and stable classifier for medical diagnosis prediction tasks.

```python
# --- Model Comparison Table ---

comparison = pd.DataFrame(results).T

comparison = comparison.round(4)

comparison = comparison.sort_values(by='Accuracy', ascending=False)

print("\nModel Comparison Table")
print("-"*35)

display(comparison)

print("\n✅ Model Comparison Table Created Successfully")
```

#### Insight:
Model comparison analysis helps evaluate the performance of multiple machine learning algorithms used for skin disorder prediction.

Comparing evaluation metrics such as Accuracy, Precision, Recall, F1 Score, and Cross Validation Score helps identify the most accurate and reliable classification model.

Sorting models based on accuracy provides a clear understanding of top-performing algorithms for medical diagnosis prediction tasks.

```python
# Feature Selection using SelectKBest

from sklearn.feature_selection import SelectKBest, f_classif

# Select Top 10 Features
selector = SelectKBest(
    score_func=f_classif,
    k=10)

# Fit on scaled data
X_selected = selector.fit_transform(X_scaled, y)

# Get selected feature names
selected_features = X.columns[selector.get_support()]

print("Selected Features")
print("-"*20)

for feature in selected_features:
    print(feature)

print("\nNumber of Selected Features :", len(selected_features))

print("\n✅ Feature Selection Completed Successfully")
```

```python
# Create Feature Scores DataFrame

feature_scores = pd.DataFrame({
    'Feature': X.columns,
    'Score': selector.scores_})

# Sort in descending order
feature_scores = feature_scores.sort_values(
    by='Score',
    ascending=False)

# Display Top 10 Features
feature_scores.head(10)
```

```python
# Create DataFrame containing selected features

X_selected_df = X[selected_features]

print("Shape of Selected Dataset :", X_selected_df.shape)

X_selected_df.head()
```

```python
# Train-Test Split

X_train_sel, X_test_sel, y_train_sel, y_test_sel = train_test_split(
    X_selected_df,
    y,
    test_size=0.2,
    random_state=42,
    stratify=y)
```

```python
results_selected = {}

for model_name, model in models.items():

    # Train
    model.fit(X_train_sel, y_train_sel)

    # Predict
    y_pred = model.predict(X_test_sel)

    # Metrics
    acc_selected = accuracy_score(y_test_sel, y_pred)

    prec_selected = precision_score(
        y_test_sel,
        y_pred,
        average='weighted',
        zero_division=0)

    rec_selected = recall_score(
        y_test_sel,
        y_pred,
        average='weighted',
        zero_division=0)

    f1_selected = f1_score(
        y_test_sel,
        y_pred,
        average='weighted',
        zero_division=0)

    # Cross Validation
    cv_score_selected = cross_val_score(
        model,
        X_selected,
        y,
        cv=5,
        scoring='accuracy').mean()

    # Train Accuracy
    train_acc_selected = accuracy_score(
        y_train_sel,
        model.predict(X_train_sel))

    # Test Accuracy
    test_acc_selected = model.score(
        X_test_sel,
        y_test_sel)

    results_selected[model_name] = {
        "Accuracy": acc_selected,
        "Precision": prec_selected,
        "Recall": rec_selected,
        "F1 Score": f1_selected,
        "CV Score": cv_score_selected,
        "Train Accuracy": train_acc_selected,
        "Test Accuracy": test_acc_selected}

print("✅ Model evaluation on selected features completed successfully.")
```

```python
# Create comparison table
results_selected_df = pd.DataFrame(results_selected).T

results_selected_df.sort_values(
    by='Accuracy',
    ascending=False,
    inplace=True)

results_selected_df
```

```python
# Compare Before and After Feature Selection
comparison_df = pd.DataFrame({

    "Dataset": [
        "All 34 Features",
        "Top 10 Features"],

    "Accuracy": [
        0.972973,
        acc_selected]})

comparison_df
```

#### Feature Selection Conclusion

Feature selection was performed using the **SelectKBest** technique with the **ANOVA F-test** to identify the top 10 most relevant features. The selected features were used to retrain all machine learning models and their performances were compared with the models trained using the complete feature set.

It was observed that the highest accuracy decreased from **97.3%** (using all 34 features) to approximately **82.4%** (using the top 10 features). This decline in performance suggests that the removed features also contribute significantly to distinguishing between different skin disorders.

Hence, feature selection did not provide any improvement in predictive performance. Therefore, the complete set of **34 features** was retained in the final model to ensure maximum accuracy and better disease classification.
